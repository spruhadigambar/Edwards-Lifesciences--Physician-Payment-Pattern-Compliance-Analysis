```python
# Changing NA records in Reciepient Name to 'Unidentified Recipient'
df['Recipient_Name'] = df['Recipient_Name'].fillna('Unidentified_Recipient')
```

```python
# Step 6 - Payment Concentration
recipient_concentration = (
    df[df['Recipient_Name'] != 'Unidentified_Recipient']
    .groupby('Recipient_Name')
    .agg(
        Payment_Records=('Payment_Record_ID', 'count'),
        Total_Payment=('Total_Amount_of_Payment', 'sum')
    )
    .sort_values('Total_Payment', ascending=False)
)

top_10_payment = recipient_concentration.head(10)['Total_Payment'].sum()

total_payment = df['Total_Amount_of_Payment'].sum()

top_10_share = (top_10_payment / total_payment) * 100

top_10_share
```

```python
# Top 10 recipients account for 37.60% of the total payment value across the entire dataset, 
# including the unidentified records.
```

```python
top_20_payment = recipient_concentration.head(20)['Total_Payment'].sum()
top_50_payment = recipient_concentration.head(50)['Total_Payment'].sum()

top_20_share = (top_20_payment / total_payment) * 100
top_50_share = (top_50_payment / total_payment) * 100

print("Top 10:", round(top_10_share, 2), "%")
print("Top 20:", round(top_20_share, 2), "%")
print("Top 50:", round(top_50_share, 2), "%")
```

```python
# Payment value is concentrated among a relatively small group of recipients, with the top 10, 20, and 50 recipients accounting for 
# 37.60%, 42.19%, and 51.34% of total payments respectively.
```

```python
concentration_summary = pd.DataFrame({
    'Recipient_Group': ['Top 10', 'Top 20', 'Top 50'],
    'Payment_Share': [top_10_share, top_20_share, top_50_share]
})

concentration_summary
```

```python
# Step 10 - Payment frequency v/s payment value
# Do recipients who receive payments frequently also account for higher total payment values?
recipient_frequency_value = df[
    df['Recipient_Name'] != 'Unidentified_Recipient'
].groupby('Recipient_Name').agg(
    Payment_Frequency=('Payment_Record_ID', 'count'),
    Total_Payment=('Total_Amount_of_Payment', 'sum')
)

recipient_frequency_value.head()
```

```python
frequency_groups = recipient_frequency_value.copy()

frequency_groups['Frequency_Group'] = pd.cut(
    frequency_groups['Payment_Frequency'],
    bins=[0, 1, 5, 20, 100, float('inf')],
    labels=['1 payment', '2-5 payments', '6-20 payments', '21-100 payments', '100+ payments']
)

frequency_summary = frequency_groups.groupby(
    'Frequency_Group',
    observed=False
).agg(
    Recipients=('Payment_Frequency', 'count'),
    Total_Payment=('Total_Payment', 'sum'),
    Average_Total_Payment=('Total_Payment', 'mean')
)

frequency_summary
```

```python
# Step 11 - Year-over-year recipient changes
recipient_yearly = df[
    df['Recipient_Name'] != 'Unidentified_Recipient'
].groupby(
    ['Recipient_Name', 'Program_Year']
).agg(
    Payment_Records=('Payment_Record_ID', 'count'),
    Total_Payment=('Total_Amount_of_Payment', 'sum')
).reset_index()

recipient_yearly.head(10)
```

```python
recipient_yearly = df[
    df['Recipient_Name'] != 'Unidentified_Recipient'
].groupby(
    ['Recipient_Name', 'Program_Year']
).agg(
    Total_Payment=('Total_Amount_of_Payment', 'sum')
).reset_index()

recipient_yoy = recipient_yearly.sort_values(
    ['Recipient_Name', 'Program_Year']
).copy()

recipient_yoy['YoY_Change'] = recipient_yoy.groupby(
    'Recipient_Name'
)['Total_Payment'].diff()

recipient_yoy['YoY_Change_%'] = recipient_yoy.groupby(
    'Recipient_Name'
)['Total_Payment'].pct_change() * 100

recipient_yoy.head(20)
```

```python
# largest year-over-year increases in actual payment dollars
largest_yoy_increases = recipient_yoy[
    recipient_yoy['YoY_Change'].notna()
].sort_values(
    'YoY_Change',
    ascending=False
)

largest_yoy_increases.head(15)
```

```python
# largest year-over-year decreases
largest_yoy_decreases = recipient_yoy[
    recipient_yoy['YoY_Change'].notna()
].sort_values(
    'YoY_Change',
    ascending=True
)

largest_yoy_decreases.head(15)
```
